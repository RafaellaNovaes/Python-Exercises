**21. Busca Binária**

```python
# lista
num = [1, 3, 5, 7, 9, 11, 13]
find = 9  # elemento para procurar

# começo e final da lista
begin = 0
end = len(num) - 1

# laço que divide a lista ao meio para achar o numero
while begin <= end:
  mid = (begin + end) // 2

  #verifica se encontrou
  if num[mid == find]:
    print("Encontrado")
    break
  elif num[mid] < find: # caso o meio for menor procurar a direita
    begin = mid + 1
  else:
    end = mid - 1       # caso maior procurar a esquerda

else:
  print("Não encontrado")
```

Para o programa cria uma lista com numeros a ser buscado que através de um laço *while* divide a lista ao meio a cada tentativa que se ajusta o início ou o fim até localizar a posição do número

**22. Criptografia de Cifra de Césa**

```python
# função definindo a cripitografia
def encrypt(msg, desloc):
  alphabet = "abcdefghijklmnopqrstuvwxyz"
  resul = "" # armazena o texto cripitografado

# percorre cada carater da mensagem
  for i in msg:
    if i in alphabet:
      position = alphabet.find(i)               # encontra a posição atual da letra
      new_position = (position + desloc) % 26   # calcula o deslocamento
      resul += alphabet[new_position]           # adiciona a nova letra ao resultado
    else:
      resul += i    # mantem os espaços sem alteração

  return resul

# entrada para o usuário digitar
msg = input("Digite uma mensagem: ")
desloc = int(input("Digite a quantidade de deslocamento: "))

print(encrypt(msg, desloc))
```

Foi criada para esse programa uma função utilizando uma string com o alfabeto. Através de um laço for, o programa localiza a posição de cada letra da mensagem com o *.find()*, soma o deslocamento e calcula uma nova posição no alfabeto, exibindo o resultado final na tela.

**23. Descriptografar Cifra de César**

```python
# função definindo a descriptografia
def decrypt(msg, desloc):
  alphabet = "abcdefghijklmnopqrstuvwxyz"
  resul = ""   # armazena o texto cripitografado

  for i in msg:
    if i in alphabet:
      position = alphabet.find(i)               # encontra a posição atual da letra
      new_position = (position - desloc) % 26   # calcula o deslocamento
      resul += alphabet[new_position]           # adiciona a nova letra ao resultado
    else:
      resul += i   # mantem os espaços sem alteração

  return resul

# entrada para o usuário digitar
msg = input("Digite a mensagem cifrada: ")
desloc = int(input("Digite a quantidade de deslocamento: "))

print(decrypt(msg, desloc))
```

Aqui foi criado uma função *decrypt* utilizando uma string com o alfabeto. Através de um laço for, o programa localiza a posição de cada letra da mensagem cifrada com o .find(), subtrai o deslocamento e calcula a posição original no alfabeto, exibindo o resultado final decifrado na tela

**24. Validação de CPF**


```python
# função defininso se o cpf é valido
def valid_cpf(cpf):
  cpf = cpf.replace(".", "").replace("-", "")

  # verifica se tem 11 digitos
  if len(cpf) != 11:
    return False

  # calcula o primeiro digito a ser verificado
  sum = 0
  weight = 10
  for i in range(9):
    sum += int(cpf[i]) * weight
    weight -= 1
  rest = sum % 11
  digit1 = 0 if rest < 2 else 11 - rest

  # calcula o segundo digito a ser verificado
  sum = 0
  weight = 11
  for i in range(10):
    sum += int(cpf[i]) * weight
    weight -= 1
  rest = sum % 11
  digit2 = 0 if rest < 2 else 11 - rest

  # confere se os dois ultimos digitos conferem com o calculo
  return cpf[-2:] == f"{digit1}{digit2}"

# entrada para o usuário
cpf = input("Digite um CPF: ")

if valid_cpf(cpf):
  print("CPF válido!")
else:
  print("CPF inválido!")

```

Para esse programa foi criado uma função chamada valid_cpf, que remove os pontos e traços do CPF e valida se ele tem 11 números que através de um laço *for*, o programa calcula o primeiro e o segundo dígito verificador multiplicando os números por pesos decrescente. Ao final, compara os dois últimos dígitos com o digito calculado infoprmando se o cpf é valido ou não

**30. Jogo da Forca**

```python
import random

# lista de palavras (strings precisam de aspas!)
words = ["telemaco", "agamenon", "penelope", "argus", "odisseu", "itaca"]
secret_word = random.choice(words)
words_find = ["_"] * len(secret_word)
tries = 6
try_words = []

print("FORCA")
print("\nPalavra:", " ".join(words_find))
print("Tentativas:", tries)

# loop principal, continua enquanto tiver tentativas e a palavra incompleta
while tries > 0 and "_" in words_find:

    # pede a letra ao jogador
    kick = input("Digite uma letra: ").lower()

    # verifica se a letra já foi tentada
    if kick in try_words:
        print("Você já tentou essa letra!")
        continue

    try_words.append(kick)  # guarda a letra tentada

    # se a letra existe, percorre e revela a posição
    if kick in secret_word:
        for i in range(len(secret_word)):
            if secret_word[i] == kick:
                words_find[i] = kick
    else:
        print("Letra errada mané!")
        tries -= 1

    # mostra o progresso a cada rodada
    print("\nPalavra:", " ".join(words_find))
    print("Tentativas:", tries)

# saída: verifica se usuário ganhou ou perdeu
if "_" not in words_find:
    print("\nParabéns, acertou a palavra:", secret_word)
else:
    print("\nVocê perdeu! A palavra era:", secret_word)
```

Foi criado para esse programa um sorteio de palavra através da lista words, onde a função *random.choice()* escolhe a *secret_word* que o jogador deve adivinhar. Através da lista *words_find*, formada por underlines (_) do mesmo tamanho da palavra secreta, o código representa as letras ainda não descobertas. Com o laço *while*, o programa repete enquanto houver tentativas (tries > 0) e ainda existir letra oculta na palavra, pedindo ao jogador que digite uma letra por vez. Através de condicionais, o código verifica se a letra já foi tentada antes (evitando repetição), se ela existe na palavra secreta (revelando sua posição em *words_find*) ou se está errada (descontando uma tentativa em *tries*). Ao final, o programa verifica se a palavra foi completamente descoberta, imprimindo a mensagem de vitória com a palavra revelada, ou a mensagem de derrota caso as tentativas acabem antes disso

**31. Ordenação por Bolha**

```python
# Lista de números
lista = [64, 34, 25, 12, 22, 11, 90]

n = len(lista)

# Algoritmo Bubble Sort
for i in range(n):
    for j in range(0, n - i - 1):
        # Se o elemento atual for maior que o próximo, eles trocam de lugar
        if lista[j] > lista[j + 1]:
            lista[j], lista[j + 1] = lista[j + 1], lista[j]

# Exibição do resultado
print("Lista ordenada:", lista)
```

Para esse programa foi criado uma lista com numeros e com dois laços *for* os percorrem para que o código percorrem os numeros vizinhos de dois em dois, se o numero atual for maior que o próximo eles trocam de posição, com esse processo ele se repete até todos os maiores valores "flutuem" para o final da lista ordenando ela inteira em ordem crescente

```python
# 1. Função da Interface do Tabuleiro
def interface():
    print("   0   1   2")
    print(
        "0  {} | {} | {}".format(
            tabuleiro[0][0], tabuleiro[0][1], tabuleiro[0][2]
        )
    )
    print("  ---+---+---")
    print(
        "1  {} | {} | {}".format(
            tabuleiro[1][0], tabuleiro[1][1], tabuleiro[1][2]
        )
    )
    print("  ---+---+---")
    print(
        "2  {} | {} | {}".format(
            tabuleiro[2][0], tabuleiro[2][1], tabuleiro[2][2]
        )
    )


# 2. Função para Validar a Vitória
def validar_vitoria(rodada):
    global parar

    # Horizontais (Linhas 0, 1 e 2)
    if (
        tabuleiro[0][0] == rodada
        and tabuleiro[0][1] == rodada
        and tabuleiro[0][2] == rodada
    ):
        interface()
        print("O '{}' venceu!".format(rodada))
        parar = True
    elif (
        tabuleiro[1][0] == rodada
        and tabuleiro[1][1] == rodada
        and tabuleiro[1][2] == rodada
    ):
        interface()
        print("O '{}' venceu!".format(rodada))
        parar = True
    elif (
        tabuleiro[2][0] == rodada
        and tabuleiro[2][1] == rodada
        and tabuleiro[2][2] == rodada
    ):
        interface()
        print("O '{}' venceu!".format(rodada))
        parar = True

    # Verticais (Colunas 0, 1 e 2)
    elif (
        tabuleiro[0][0] == rodada
        and tabuleiro[1][0] == rodada
        and tabuleiro[2][0] == rodada
    ):
        interface()
        print("O '{}' venceu!".format(rodada))
        parar = True
    elif (
        tabuleiro[0][1] == rodada
        and tabuleiro[1][1] == rodada
        and tabuleiro[2][1] == rodada
    ):
        interface()
        print("O '{}' venceu!".format(rodada))
        parar = True
    elif (
        tabuleiro[0][2] == rodada
        and tabuleiro[1][2] == rodada
        and tabuleiro[2][2] == rodada
    ):
        interface()
        print("O '{}' venceu!".format(rodada))
        parar = True

    # Diagonais
    elif (
        tabuleiro[0][0] == rodada
        and tabuleiro[1][1] == rodada
        and tabuleiro[2][2] == rodada
    ):
        interface()
        print("O '{}' venceu!".format(rodada))
        parar = True
    elif (
        tabuleiro[2][0] == rodada
        and tabuleiro[1][1] == rodada
        and tabuleiro[0][2] == rodada
    ):
        interface()
        print("O '{}' venceu!".format(rodada))
        parar = True


# Matriz do tabuleiro (com índices 0, 1 e 2)
tabuleiro = [[" ", " ", " "], [" ", " ", " "], [" ", " ", " "]]

parar = False
rodada = "X"
jogadas = 0

# Laço Principal do Jogo
while parar == False:
    interface()

    linha = int(input("Digite a linha escolhida (0-2): "))
    coluna = int(input("Digite a coluna escolhida (0-2): "))

    if rodada == "X":
        tabuleiro[linha][coluna] = "X"
        jogadas += 1
        validar_vitoria(rodada)
        rodada = "O"
    elif rodada == "O":
        tabuleiro[linha][coluna] = "O"
        jogadas += 1
        validar_vitoria(rodada)
        rodada = "X"

    # Verificação de Empate
    if jogadas == 9 and parar == False:
        interface()
        print("Empate!")
        parar = True

print("Programa encerrado.")
```

Para esse programa, foi criada uma tabuleiro 3x3 que com um laço *while*, os jogadores alternam jogadas digitando a linha e a coluna desejadas. A cada rodada, o código confere se alguém completou uma linha, coluna ou diagonal para declarar o vencedor, ou encerra em empate caso chegue a 9 jogadas

**35. Números Fibonacci Até N**

```python
# Números Fibonacci até N

n = int(input("Digite o valor de N: "))

a, b = 0, 1

print("Sequência de Fibonacci até", n, ":")

# repete enquanto o valor atual (a) não ultrapassar n
while a <= n:
    print(a, end=" ")
    a, b = b, a + b
```

Foi criado para esse programa um laço *while* que gera a sequência de Fibonacci até um número n fornecido pelo usuário, usando as variáveis a e b para guardar o número atual e o próximo. Enquanto a <= n, o código imprime o valor atual e atualiza os dois números com a, b = b, a + b, avançando na sequência. O laço se repete até que a ultrapasse n, encerrando a geração da sequência

```python
# lista vazia
shop_list = []

# menu de opções
while True:
  print("\n--- LISTA DE COMPRAS ---")
  print("1 - Adicionar item")
  print("2 - Remover item")
  print("3 - Listar itens")
  print("4 - Sair")

  op = input("Escolha uma opção: ")

# condicionais das opções
  if op == "1":
      item = input("Digite o nome do item: ")
      shop_list.append(item)
      print(f"'{item}' adicionado com sucesso!")

  elif op == "2":
      item = input("Digite o item que deseja remover: ")
      if item in shop_list:
          shop_list.remove(item)
          print(f"'{item}' removido com sucesso!")
      else:
          print("Item não encontrado na lista.")

  elif op == "3":
      if shop_list:
          print("\nItens na lista:")
          for i, item in enumerate(shop_list, start=1):
              print(f"{i}. {item}")
      else:
          print("A lista está vazia.")

  elif op == "4":
      print("Saindo do programa")
      break

  else:
      print("Opção inválida, tente novamente.")
```

**48. Ordenação por Inserção**

```python
nums = [12, 15, 18, 5, 6]

# percorre a lista apartir do segundo elemento
for i in range(1, len(nums)):
  key = nums[i]
  j = i - 1

  # desloca para a direita os numeros maiores
  while j >= 0 and nums[j] > key:
    nums[j + 1] = nums[j]
    j -= 1

  # insere o numero na posição correta (insertion sort)
  nums[j + 1] = key

# saida
print(nums)
```

O programa percorre a lista a partir do segundo elemento guardando o valor atua na variável *key* que desloca os numeros dos menores para o maior ate ordernar a lista

**50. Cálculo de Frequência de Letras**

```python
text = input("Digite uma string: ")
freq = {}

# percorre cada caracter do texto
for i in text:
  # verifica se o caracter é letra ignorando numeros e simbolos
  if i.isalpha():
    i = i.lower()
    # contador do dicionario
    if i in freq:
      freq[i] += 1
    else:
      freq[i] = 1

# saida
for i, quant in freq.items():
    print(f"{i}: {quant}")
```

o programa recebe uma string do usuário que com o laço *for* percorre cada caracter, com o método *isalpha()* filtra apenas letras que com o método *.lower()* converte para letras minúsculas. A contagem de cada letra é acumulada no dicionário e exibida no final
